**O(1) < O(logn) < O(n) < O(n log n) < O(n^2) < O(2^n) < O(n!)**

```python
#LINEAR O(n)
n =10
for i in range(n):
    print(i,end=" ")
```

```python
# QUADRETIC o(n^2)
for i in range(n**2):
    print(i,end=" ")
```

```python
#LOGARITHMIC O(logn)
for i in range(n):
    if i%2==0:
        print(i,end=" ")
```
